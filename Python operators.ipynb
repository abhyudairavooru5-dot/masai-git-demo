# Bitwise Operators

```python
# used to compare binary numbers
```

## & (and) set each bit to 1 if both bits are 1
## | (or) set each bit to 1 if either is 1
## ^ (xor) set each bit to 1 if only one bit is 1
## ~ (not) invert all the bits
## << zero fill left shift: shift left by pushing zeroes in from the right and let the left most bits fall off
## >> signed right shift shift right by losing copies of the left most bits in form of left and let the right most bits fall off

## Number         Binary
## 1               1
## 2               10
## 3               11
## 4               100
## 5               101
## 6               110
## 7               111
## 8               1000
## 9               1001
## 10              1010

```python
x = 6
y = 3
# 6 = 110
# 3 = 011
# 2 = 010
```

```python
print(x & y)
```

```python
print(x | y)
# 6 = 110
# 3 = 011
# 7 = 111
```

```python
print(x ^ y)
# 6 = 110
# 3 = 011
# 5 = 101
```

```python
print(~ x)
# 6 = 110
#   = 001
```

```python
print(x << 3)
```

```python
print(x << 2)
```

```python
print(x >> 2)
```
