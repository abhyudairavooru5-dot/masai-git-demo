```python
def calculator (a, b, ops):
  if ops == "add":
    return a + b
  elif ops == "sub":
    return a - b
  elif ops == "mul":
    return a * b
  elif ops == "div":
    if b == 0:
      print("Error: Cannot divide by zero.")
      return None
    return a / b
  elif ops == "mod":
    return a % b
  elif ops == "percentage":
    return a / b*100
  else:
    print ("Please use a valid operation among: 'add', 'sub', 'mul', 'div', 'mod', 'percentage'")
def main ():
  print ("Enter the value (a):")
  a = int (input())
  print ("Enter the value (b):")
  b = int (input())
  print ("Enter the operation among: 'add', 'sub', 'mul', 'div', 'mod', 'percentage'")
  ops = input()
  print (calculator (a, b, ops))

main ()
```
